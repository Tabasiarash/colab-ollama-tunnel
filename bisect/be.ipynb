```python
import socket, subprocess, datetime, json
res = {}
res['time'] = str(datetime.datetime.utcnow())
try:
    r = subprocess.run(['nvidia-smi'], capture_output=True, text=True)
    res['nvidia_smi_rc'] = r.returncode
    res['nvidia_head'] = (r.stdout or r.stderr)[:400]
except Exception as e:
    res['nvidia_smi_exc'] = repr(e)
for host in ['api.github.com','ollama.com','api.cloudflare.com','google.com']:
    try:
        res['resolve_'+host] = socket.gethostbyname(host)
    except Exception as e:
        res['resolve_'+host] = 'FAIL ' + repr(e)
try:
    r = subprocess.run(['curl','-s','-o','/dev/null','-w','%{http_code}','--max-time','15','https://api.github.com'], capture_output=True, text=True)
    res['curl_api_github'] = (r.stdout, r.stderr[:200])
except Exception as e:
    res['curl_exc'] = repr(e)
try:
    r = subprocess.run(['curl','-s','-o','/dev/null','-w','%{http_code}','--max-time','15','https://ollama.com'], capture_output=True, text=True)
    res['curl_ollama'] = (r.stdout, r.stderr[:200])
except Exception as e:
    res['curl_exc2'] = repr(e)
print(json.dumps(res, indent=2), flush=True)
with open('/kaggle/working/env.txt','w') as f:
    f.write(json.dumps(res, indent=2)+'\n')
print('env.txt written', flush=True)
```
